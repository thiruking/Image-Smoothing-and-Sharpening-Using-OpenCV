```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("d1.webp")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python

avg_filter = cv2.blur(img_rgb, (5,5))
```

```python

kernel = np.array([[1,2,1],
                   [2,4,2],
                   [1,2,1]]) / 16

weighted_avg = cv2.filter2D(img_rgb, -1, kernel)
```

```python
gaussian_filter = cv2.GaussianBlur(img_rgb, (5,5), 0)
```

```python

median_filter = cv2.medianBlur(img_rgb, 5)
```

```python
laplacian_kernel = np.array([[0,-1,0],
                             [-1,5,-1],
                             [0,-1,0]])

laplacian_sharp = cv2.filter2D(img_rgb, -1, laplacian_kernel)
```

```python
gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

laplacian_operator = cv2.Laplacian(gray, cv2.CV_64F)

laplacian_operator = np.uint8(np.absolute(laplacian_operator))
```

```python
plt.figure(figsize=(15,12))


# Original Image
plt.subplot(3,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")


plt.subplot(3,3,2)
plt.imshow(avg_filter)
plt.title("Averaging Filter")
plt.axis("off")
```

```python
plt.figure(figsize=(12,15))


# Original Image
plt.subplot(3,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

# Weighted Averaging
plt.subplot(3,3,2)
plt.imshow(weighted_avg)
plt.title("Weighted Averaging")
plt.axis("off")
```

```python
plt.figure(figsize=(12,15))


# Original Image
plt.subplot(3,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(3,3,2)
plt.imshow(gaussian_filter)
plt.title("Gaussian Filter")
plt.axis("off")
```

```python
plt.figure(figsize=(12,15))


# Original Image
plt.subplot(3,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(3,3,2)
plt.imshow(median_filter)
plt.title("Median Filter")
plt.axis("off")
```

```python
plt.figure(figsize=(12,15))


# Original Image
plt.subplot(3,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(3,3,2)
plt.imshow(laplacian_sharp)
plt.title("Laplacian Sharpening")
plt.axis("off")


# Laplacian Operator
plt.subplot(3,3,3)
plt.imshow(laplacian_operator, cmap='gray')
plt.title("Laplacian Operator")
plt.axis("off")
```
